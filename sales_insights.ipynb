```python
#loading the data
import pandas as pd
df = pd.read_csv("superstore_analysis.csv", encoding="ISO-8859-1")
df.drop(columns=['Column1'], inplace=True)
print(df.head())
```

```python
df.info()
```

```python
df.describe()
```

```python
#handling the missing values
df.isnull().sum()
```

```python
#cleaning the column names
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
print(df.head(10))
```

```python
#Changing the data types
df[['order_date','ship_date']] = df[['order_date','ship_date']].apply(pd.to_datetime, errors ="coerce")
```

```python
df.info()
```

```python
#checking if their is any duplicate value in dataset
df.duplicated().any()
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(keep=False)]
```

```python
#dropping the duplicate item
df = df.drop_duplicates()
```

```python
df.duplicated().any()
```

```python
df['year']=df['order_date'].dt.year
df['month']=df['order_date'].dt.month
df['month_name']=df['order_date'].dt.month_name()
```

```python
df['profit_margin'] = df['profit'] / df['sales']
```

```python
df.head()
```

```python
#checking the outliers in the dataset
import numpy as np
q1 = df['sales'].quantile(0.25)
q3 = df['sales'].quantile(0.75)
iqr = q3 - q1
df['is_outlier'] = ((df['sales'] < (q1 - 1.5*iqr)) | (df['sales'] > (q3 + 1.5*iqr)))
```

```python
df.head()
```

```python
df.to_csv("cleaned_superstore.csv", index=False)
```
